# Theta*: Any-Angle Path Planning on Grids

A* on a grid builds paths from moves between neighboring cells, restricting their directions. **Theta\*** still generates neighbors on the grid, but allows a node's parent to be farther away when the two nodes have **line of sight (LOS)**. The resulting path can contain straight segments at arbitrary angles. In this lab, nodes represent cell centers and segment costs are Euclidean distances.

When expanding a node $s$, consider each free neighbor $n$. If $s$ has a parent $p$ and the segment from $p$ to $n$ is clear, try connecting $n$ directly to $p$:

$$
\tilde g(n) = g(p) + d(p,n).
$$

Otherwise, connect $n$ to $s$, as in A*:

$$
\tilde g(n) = g(s) + d(s,n).
$$

Here, $g(x)$ is the cost of the path to $x$, and $d(a,b)$ is Euclidean distance. Use the candidate connection when it improves the known cost of reaching $n$. The search keeps A*'s OPEN/CLOSED structure and uses Euclidean distance to estimate the remaining cost. Following parent pointers reconstructs the path, including its longer straight segments.

Theta* often finds shorter, straighter paths than grid A*, but **does not guarantee the shortest possible path**. It only considers the current node and its parent as possible connections during this update, so it can miss a better connection through another visible node.
